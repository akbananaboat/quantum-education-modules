# Interference

## 1. Interference in Classical Physics


### 1.1 Interference in Waves
Waves are a crucial part of classical physics, and they are defined as distrubances that travel through a medium. For example, sound travels as a wave with air as its medium, and ripples in a pond are waves that use water as its medium. Waves have high points called crests and low points called troughs. When waves interact, they combine to either amplify or cancel out, and this is called interference. 

When waves amplify each other, it is called constructive interference. This occurs when two crests or two troughs overlap. In the example below, the crest of the red wave and blue waves are perfectly align, and they result in the black wave.

## 2. Interference in Quantum

In a quantum setting, interference is slightly different. Instead of looking at physical waves interacting, quantum looks at the wave like behavior of particles. In quantum physics, particles are represented by wave functions. The wave function represents the possible paths a particle could take, and the probability that each path occurs is the probability amplitude. When two waves interact, their probabilty amplitudes either interact constructively or destructively. 

When the waves interact constructively, the odds of the particle taking a particular path increase. When they interact destructively, then it is less likely for a particle to take that path.

An important thing to note is that the concept of a wave function only exists when a particle is in superposition. The moment a state of a particle is measured, the wave function collapses to a definite state.
